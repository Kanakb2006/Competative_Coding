```python
#wap to aacept three paper marks and check maximum marks using nested if else
#Nested if else:
n1 = int (input("Enter the marks of paper 1: "))
n2 = int (input("Enter the marks of paper 2: "))
n3 = int (input("Enter the marks of paper 3: "))
if n1>n2:
    if n1>n3:
        print("Paper 1 has the maximum marks: ", n1)
    else:
        print("Paper 3 has the maximum marks: ", n3)
else:    
    if n2>n3:
        print("Paper 2 has the maximum marks: ", n2)
    else:       
     print("Paper 3 has the maximum marks: ", n3)
```

```python
mydict = {
    101 : "kanak",
    102 : "developer",
    "103":"manager",
    "104":"kohana"
}
print(mydict)
# with the help of key we have to print values
# a= mydict[102]
#print(a)

# we will replace old value by new value
mydict[102] = "tester"
print(mydict)

## only print key x=0,1
for x in mydict:
    print(x)
## only print value
for x in mydict:
    print(mydict[x])
# printing  key and values both :
for x,y in mydict.items():
    print(x,y)
# if i have to add new key and value in dictionary
mydict["mobile.no"] = 1234567890
print(mydict)
    
```
