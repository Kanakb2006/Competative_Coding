```python
# wap to accept three paper marks and calculate total percentage if greater than equal to 60 then he/she is eligible for placement
math = int(input("Enter the marks of math: "))
phy = int(input("Enter the marks of physics: "))
chem = int(input("Enter the marks of chemistry: "))
total = math + phy + chem
percentage = (total/300)*100
print("The percentage is: ", percentage)
if percentage >= 60:
    print("He/She is eligible for placement")
else:
    print("He/She is not eligible for placement")
    
```
