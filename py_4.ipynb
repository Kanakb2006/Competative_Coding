```python
# How to return multiple value
def add(): # called function
    n1 = int(input("Enter the value of n1: "))
    n2 = int(input("Enter the value of n2: "))
    sum = n1+n2
    mul = n1*n2
    sub = n1-n2
    div = n1/n2
    
    return  sum, mul, sub, div
result = add()
print(result)
```

```python
# four types of argument
#1. positional argument
def personalInfo(fname, lname):
    print("First Name:", fname)
    print("Last Name:", lname)
    personalInfo("Kohana", "aditi") #positional argument
#2.keyword argument
def personalInfo(fname, lname):
    print("First Name:", fname)
    print("Last Name:", lname)
    personalInfo(fname="Kohana", lname="aditi") #keyword argument
#3.default argument
def cityName(city = "Mumbai"):
    print(city)
cityName("Pune") #default argument
cityName("Delhi")
cityName()
#4.variable length argument/variable number of argument
def studentNames(*name):
    print(name)
studentNames("Kohana", "Aditi", "Janhavi", "Kanak")
    
```

```python
mylist = [1, 2, 3, 4, 5]
# search the element 7
def searchElement(target):
    for i in range(len(mylist)): #mylist = 6
        if target == mylist[i]: #target = 7, mylist[0] = 1
            print("Element found at index", i)
    return -1   
result = searchElement(4) # 7 this element we have to search
if result != -1:
    print("Element found at index", result)
else:
    print("Element not found in the list")
```
