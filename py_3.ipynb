```python
name = "kanak"
newname = ""
for i in name:
    if i not in newname:
        newname += i
print(name)
print(newname)
```

```python
# check for palindrome string
name = "madam"
if name == name[::-1]:
    print("This is a palindrome")
else:    print("This is not a palindrome")
```

```python
# check for Anagrams
name1 = "listen"
name2 = "silent"
if sorted(name1) ==  sorted (name2):
    print("This is an anagram")
else:
    print("This is not an anagram")
```

```python
# Add key value pair in dictionary
mydict = {"name":"kanak","age":19}
mydict["gender"] = "female"
print(mydict)   
```

```python
# check if a key exists in a dictionary
mydict = {"name":"kanak","age":19,"gender":"female"}
if "name" in mydict:
    print("Key exists in the dictionary")
else:    print("Key does not exist in the dictionary")
```
