```python
message = input()

whitespace_count = 0
special_count = 0

for ch in message:
    if ch.isspace():
        whitespace_count += 1
    elif not ch.isalnum():
        special_count += 1

print("Whitespace:", whitespace_count)
print("Special characters:", special_count)
```

```python
A=[1,2,3]
B=[2,3,4]
C=[3,4,5]
for i in A :
    if i in B and i in C:
        print(i)
```

```python
#REMOVE ZEROS FROM THE ARRAY AND MOVES THEM TO THE END
a=[1,0,3,4,5]
for i in a:
    if i==0:
        a.remove(i)
        a.append(0)
print(a)
```

```python
#Finding the total distance between adjacent items of a list of 5 numbers
b=[1,2,3,4,5]
total_distance = 0
for i in range(len(b)-1):
    total_distance += abs(b[i] - b[i+1])    
print(total_distance)
```
